```python
import os
from PIL import Image
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
# ========================  inputs  ======================================
folder = r''
request = ""

sample_1 = ''
sample_2 = ''
sample_3 = ''
sample_4 = ''

# first_mag = '100x'
# second_mag = '2kx'
# third_mag = '10kx'
# fourth_mag = '50kx'

first_mag = ''
second_mag = ''
third_mag = ''
fourth_mag = ''

samples_are_columns = False          # denote if samples are columns of rows
samples_are_grid_pos = False        # if each sample has just one image, ***above must be false***
file_name_included = True           # set as True or False
# ======================================================================
mag_scales = { '250kx': '20 nm','200kx': '50 nm', '150kx': '50 nm', '100kx': '100 nm', '80kx': '100 nm', '60kx': '100 nm', 
               '50kx' :'200 nm', '40kx': '200 nm', '30kx': '200 nm', '25kx': '200 nm', '20kx': '500 nm', '15kx': '500 nm', 
               '12kx' :'500 nm', '10kx': '1 μm',    '8kx': '1 μm',    '6kx': '1 μm',    '5kx': '2 μm',    '4kx': '2 μm',    
                 '3kx': '2 μm', '2.5kx': '2 μm',    '2kx': '5 μm',   '100x': '50 μm', '':''}

first_scale = mag_scales[first_mag] 
second_scale = mag_scales[second_mag]
third_scale = mag_scales[third_mag]
fourth_scale = mag_scales[fourth_mag]

def create_tem_grid(image_folder, output_pdf):
    c = canvas.Canvas(output_pdf, pagesize=letter)
    page_width, page_height = letter
    
    valid_extensions = ('.png', '.jpg', '.jpeg', '.tiff', '.tif', '.TIFF')
    all_files = sorted(os.listdir(image_folder))
    image_files = [f for f in all_files if f.lower().endswith(valid_extensions)]
    grid_positions = []
    file_count = len(image_files)
# ================================={ 4 image setup }=================================================
    if file_count == 4:
        col_x = [75, 340] 
        row_y = [500, 300]    
        box_w = 246
        box_h = 190
        
        if samples_are_columns:
            # Add a main page title at the top
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 750, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 700, sample_1)
            c.drawString(415, 700, sample_2)
            c.drawString(25, 600, first_mag)
            c.drawString(25, 400, second_mag)
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 530, first_scale)
            c.drawString(545, 530, first_scale)
            c.drawString(280, 330, second_scale)
            c.drawString(545, 330, second_scale)
            grid_positions = [
            (0, 0),  # Image 1: Left Column, Top Row
            (0, 1),  # Image 2: Left Column, Bottom Row
            (1, 0),  # Image 4: Right Column, Top Row
            (1, 1),  # Image 5: Right Column, Bottom Row
            ]
        else:
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 750, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 700, first_mag)
            c.drawString(415, 700, second_mag)
            c.drawString(25, 600, sample_1)
            c.drawString(25, 400, sample_2)
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 530, first_scale)
            c.drawString(545, 530, second_scale)
            c.drawString(280, 330, first_scale)
            c.drawString(545, 330, second_scale)
            grid_positions = [
            (0, 0),  # Image 1: Left Column, Top Row
            (1, 0),  # Image 2: Left Column, Bottom Row
            (0, 1),  # Image 4: Right Column, Top Row
            (1, 1),  # Image 5: Right Column, Bottom Row
            ]
# ================================={ 6 image setup }=================================================
    elif file_count == 6 or file_count == 5:
        col_x = [75, 340] 
        row_y = [500, 300, 100]    
        box_w = 246
        box_h = 190
    
        if samples_are_columns:
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 750, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 700, sample_1)
            c.drawString(415, 700, sample_2)
            c.drawString(25, 600, first_mag)
            c.drawString(25, 400, second_mag)
            c.drawString(25, 200, third_mag)
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 530, first_scale)
            c.drawString(545, 530, first_scale)
            c.drawString(280, 330, second_scale)
            c.drawString(545, 330, second_scale)
            c.drawString(280, 130, third_scale)
            if file_count == 6:
                c.drawString(545, 130, third_scale)
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (0, 1),  # Image 2: Left Column, Middle Row
                (0, 2),  # Image 3: Left Column, Bottom Row
                (1, 0),  # Image 4: Right Column, Top Row
                (1, 1),  # Image 5: Right Column, Middle Row
                (1, 2)   # Image 6: Right Column, Bottom Row
                ]
            elif file_count == 5:
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (0, 1),  # Image 2: Left Column, Middle Row
                (0, 2),  # Image 3: Left Column, Bottom Row
                (1, 0),  # Image 4: Right Column, Top Row
                (1, 1),  # Image 5: Right Column, Middle Row
                ]
        elif samples_are_grid_pos:
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 750, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 700, sample_1)
            c.drawString(415, 700, sample_1)
            c.drawString(25, 600, first_mag)
            c.drawString(25, 400, first_mag)
            c.drawString(25, 200, first_mag)
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 530, first_scale)
            c.drawString(545, 530, first_scale)
            c.drawString(280, 330, first_scale)
            c.drawString(545, 330, first_scale)
            c.drawString(280, 130, first_scale)
            if file_count == 6:
                c.drawString(545, 130, first_scale)
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (0, 1),  # Image 2: Left Column, Middle Row
                (0, 2),  # Image 3: Left Column, Bottom Row
                (1, 0),  # Image 4: Right Column, Top Row
                (1, 1),  # Image 5: Right Column, Middle Row
                (1, 2)   # Image 6: Right Column, Bottom Row
                ]
            elif file_count == 5:
                c.drawString(545, 130, first_scale)
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (0, 1),  # Image 2: Left Column, Middle Row
                (0, 2),  # Image 3: Left Column, Bottom Row
                (1, 0),  # Image 4: Right Column, Top Row
                (1, 1),  # Image 5: Right Column, Middle Row
                ]
        else:
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 750, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 700, first_mag)
            c.drawString(415, 700, second_mag)
            c.saveState()
            c.translate(65,180)
            c.rotate(90)
            c.drawString(400, 0, sample_1)
            c.drawString(200, 0, sample_2)
            c.drawString(0, 0, sample_3)
            c.restoreState()
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 530, first_scale)
            c.drawString(545, 530, second_scale) 
            c.drawString(280, 330, first_scale)
            c.drawString(545, 330, second_scale)
            c.drawString(280, 130, first_scale)
            if file_count == 6:
                c.drawString(545, 130, second_scale)
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (1, 0),  # Image 2: Left Column, Middle Row
                (0, 1),  # Image 3: Left Column, Bottom Row
                (1, 1),  # Image 4: Right Column, Top Row
                (0, 2),  # Image 5: Right Column, Middle Row
                (1, 2)   # Image 6: Right Column, Bottom Row
                ]
            elif file_count == 5:
                grid_positions = [
                    (0, 0),  # Image 1: Left Column, Top Row
                    (1, 0),  # Image 2: Left Column, Middle Row
                    (0, 1),  # Image 3: Left Column, Bottom Row
                    (1, 1),  # Image 4: Right Column, Top Row
                    (0, 2),  # Image 5: Right Column, Middle Row
                    ]
# ================================={ 8 image setup }=================================================
    elif file_count == 8 or file_count == 7:
        col_x = [75, 340] 
        row_y = [550, 367, 183, 0]    
        box_w = 246
        box_h = 190


        if samples_are_columns:
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 760, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 740, sample_1)
            c.drawString(415, 740, sample_2)
            c.drawString(25, 660, first_mag)
            c.drawString(25, 466, second_mag)
            c.drawString(25, 283, third_mag)
            c.drawString(25, 100, fourth_mag)
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 577, first_scale)
            c.drawString(545, 577, first_scale)
            c.drawString(280, 393, second_scale)
            c.drawString(545, 393, second_scale)
            c.drawString(280, 210, third_scale)
            c.drawString(545, 210, third_scale)
            c.drawString(280, 27, fourth_scale)
            if file_count == 8:
                c.drawString(545, 27, fourth_scale)
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (0, 1),  # Image 2: Left Column, Top Middle Row
                (0, 2),  # Image 3: Left Column, Bottom Middle Row
                (0, 3),  # Image 4: Left Column, Bottom Row
                (1, 0),  # Image 5: Right Column, Top Row
                (1, 1),  # Image 6: Right Column, Top Middle Row
                (1, 2),  # Image 7: Right Column, Bottom Middle Row
                (1, 3)   # Image 8: Right Column, Bottom Row
                ]
            elif file_count == 7:
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (0, 1),  # Image 2: Left Column, Top Middle Row
                (0, 2),  # Image 3: Left Column, Bottom Middle Row
                (0, 3),  # Image 4: Left Column, Bottom Row
                (1, 0),  # Image 5: Right Column, Top Row
                (1, 1),  # Image 6: Right Column, Top Middle Row
                (1, 2),  # Image 7: Right Column, Bottom Middle Row
                ]
        else:
            c.setFont("Helvetica-Bold", 24)
            c.drawString(25, 760, request)
            c.setFont("Helvetica-Bold", 16)
            c.drawString(165, 740, first_mag)
            c.drawString(415, 740, second_mag)
            c.saveState()
            c.translate(55,80)
            c.rotate(90)
            c.drawString(549, 0, sample_1)
            c.drawString(378, 0, sample_2)
            c.drawString(183, 0, sample_3)
            c.drawString(0, 0, sample_4)
            c.restoreState()
            c.setFont("Helvetica-Bold", 12)
            c.drawString(280, 577, first_scale)
            c.drawString(545, 577, second_scale)
            c.drawString(280, 393, first_scale)
            c.drawString(545, 393, second_scale)
            c.drawString(280, 210, first_scale)
            c.drawString(545, 210, second_scale)
            c.drawString(280, 27, first_scale)
            if file_count == 8:
                c.drawString(545, 27, second_scale)
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (1, 0),  # Image 2: Left Column, Top Middle Row
                (0, 1),  # Image 3: Left Column, Bottom Middle Row
                (1, 1),  # Image 4: Left Column, Bottom Row
                (0, 2),  # Image 5: Right Column, Top Row
                (1, 2),  # Image 6: Right Column, Top Middle Row
                (0, 3),  # Image 7: Right Column, Bottom Middle Row
                (1, 3)   # Image 8: Right Column, Bottom Row
                ]
            elif file_count == 7:
                grid_positions = [
                (0, 0),  # Image 1: Left Column, Top Row
                (1, 0),  # Image 2: Left Column, Top Middle Row
                (0, 1),  # Image 3: Left Column, Bottom Middle Row
                (1, 1),  # Image 4: Left Column, Bottom Row
                (0, 2),  # Image 5: Right Column, Top Row
                (1, 2),  # Image 6: Right Column, Top Middle Row
                (0, 3),  # Image 7: Right Column, Bottom Middle Row
                ]
# ================================================================================================
        # 5. Loop and draw images into the mapped spots
    if file_count in [4, 5, 6, 7, 8]:
        for i, filename in enumerate(image_files):
            image_path = os.path.join(image_folder, filename)
            col, row = grid_positions[i]
            
            x = col_x[col]
            y = row_y[row] + 20
    
            with Image.open(image_path) as img:
                img_w, img_h = img.size
            ratio = min(box_w / img_w, box_h / img_h)
            display_w = img_w * ratio
            display_h = img_h * ratio
            
            x_centered = x + (box_w - display_w) / 2
            y_centered = y + (box_h - display_h) / 2
    
            c.drawImage(image_path, x_centered, y_centered, width=display_w, height=display_h, preserveAspectRatio=True)
            
            c.setFont("Helvetica-Bold", 10)
            if file_name_included:
                c.drawString(x, y, f"{filename[:25]}")
        c.save()
        print(f"Success! Grid PDF saved as: {output_pdf}")
    else:
        print(f'Incompatible file amount. There should be 2, 4, 6, 8 or 10, there are {len(image_files)} files')


create_tem_grid(folder, output_pdf=f"{folder}\{request}_TEM_Image_Report_.pdf")
```
